```python
#Un sistema de transporte otorga tarifa reducida si el usuario es estudiante o es un adulto mayor (60 años o más)
Estudiante=str(input("¿Eres estudiante? (si/no):"))
Edad=int(input("Ingrese su edad:"))
respEstudiante=(Estudiante=="si")
if respEstudiante:
  print(f"Usted tiene derecho a una tarifa reducida")
if Edad>=60:
  print(f"Usted tiene derecho a una tarifa reducida")
else:
  print(f"Usted no tiene derecho a una tarifa reducida")

```

```python
a=7
b=8
c=9

semíperimetro=(a+b+c)/2
area=(semíperimetro*(semíperimetro-a)*(semíperimetro-b)*(semíperimetro-c))**0.5

print(f"El semiperimetro del triangulo es: {semíperimetro}")
print(f"El area es:{area}")
```
