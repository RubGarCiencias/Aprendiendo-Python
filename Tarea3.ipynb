```python
#Ejercicio 1
#Programa para determinar si una recta es paralela ,perpendicular o coincidente

m=float(input("ingrese el valor de la pendiente de la primera recta: "))
b=float(input("ingrese el valor de la ordenada al origen de la primera recta: "))

m2=float(input("ingrese el valor de la pendiente de la segunda recta: "))
b2=float(input("ingrese el valor de la ordenada al origen de la segunda recta: "))

if m==m2 and b==b2:
  print(f"Las rectas son coincidentes")
elif m == m2 :
  print(f"Las rectas son paralelas")
elif m*(1/m2)==-1:
  print(f"Las rectas son perpendiculares")
else:
  print(f"Las rectas no son paralelas ,perpendiculares o coincidentes")
```

```python
#Ejercicio 3: Desigualdad triangular y formula de herón
a=float(input("ingrese el valor del vertice a: "))
b=float(input("ingrese el valor del vertice b: "))
c=float(input("ingrese el valor del vertice c: "))
s=(a+b+c)/2
area=(s*(s-a)*(s-b)*(s-c))**0.5

if a+b<c or b+c<a or a+c<b:
  print("Los vertices no forman un triangulo, ingrese otros valores")

elif a==b==c:
  print("El triangulo es equilatero")
  print(f"El area del triangulo es: {area}")

elif a==b or b==c or a==c:
  print("El triangulo es isoceles")
  print(f"El area del triangulo es: {area}")

else:
  print("El triangulo es escaleno")
  print(f"El area del triangulo es: {area}")
```

```python
#Ejercicio 9: Volumen y Área de la Superficie de una Pirámide Regular Recta
import math
h=float(input("Ingrese la altura de la pirámide: "))
n=int(input("Ingrese la cantidad de lados del poligono que constituye la base: "))
s=float(input("Ingrese la longitud de los lados del poligono: "))
a=(s/2)*(1/math.tan(math.pi/n))
l=math.sqrt(h**2+a**2)
A=(n*s*a)/2

V=(A*h)/3
S=A+((n*s*l)/2)

if h>0 and n>=3 and s>0:
  print(f"El volumen de la pirámide es: {V} cm^3")
  print(f"El area de la superficie de la pirámide es: {S} cm^2" )

else:
  print("Error, ingrese valores distintos")
```
